```python
# Graph Building (User Input)
d = {}
n = int(input("Enter total nodes: "))
for i in range(n):
    u = input(f"Enter node {i}: ")
    d[u] = []
    h = int(input(f"Enter number of neighbours at node {i}: "))
    for j in range(h):
        v = input()
        d[u].append(v)
d
```

```python
# DFS - Recursive (returns True/False)
def dfs(st, goal, graph, vis=None):
    if vis is None:
        vis = set()
    print(f"Visiting: {st}")
    if st == goal:
        return True
    vis.add(st)
    for child in graph.get(st, []):
        if child not in vis:
            if dfs(child, goal, graph, vis):
                return True
    return False

# 🆕 Different Data — City Road Network
graph = {
    'Mumbai':   ['Pune', 'Surat'],
    'Pune':     ['Mumbai', 'Hyderabad', 'Nagpur'],
    'Surat':    ['Mumbai', 'Ahmedabad'],
    'Ahmedabad':['Surat', 'Jaipur'],
    'Hyderabad':['Pune', 'Chennai', 'Nagpur'],
    'Nagpur':   ['Pune', 'Hyderabad', 'Delhi'],
    'Chennai':  ['Hyderabad', 'Delhi'],
    'Delhi':    ['Nagpur', 'Chennai', 'Jaipur'],
    'Jaipur':   ['Ahmedabad', 'Delhi']
}

result = dfs('Mumbai', 'Jaipur', graph)
print(f"\nPath found: {result}")
```

```python
# DFS - Iterative (returns full path taken)
def dfs_iterative(s, goal, graph):
    res = []
    stack = []
    vis = set()
    stack.append(s)
    while len(stack) != 0:
        u = stack.pop()
        if u not in vis:
            vis.add(u)
            res.append(u)
            if u == goal:
                return res
            for v in graph[u]:
                stack.append(v)
    return res

result2 = dfs_iterative('Mumbai', 'Jaipur', graph)
print(f"Path found: {result2}")
```
