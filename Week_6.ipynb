```python
import networkx as nwx
```

```python
graph = nwx.Graph()
```

```python
graph
```

```python
while True:
    choice = int(input("Want to insert nodes and enter edge press? 1 for yes and 2 for no"))
    if choice == 1:
        node1 = input("Enter node1 ").strip().upper()
        node2 = input("Enter node2 ").strip().upper()
        weight = int(input(f"Enter weight between {node1} and {node2} "))
        graph.add_edge(node1, node2, weight = weight)
    else:
        break
```

```python
graph.adj.items()
```

```python
for node, neighbors in graph.adj.items():
    print(node, neighbors)
```

```python
start = input("Enter start node").strip().upper()
goal = input("Enter goal node").strip().upper()
```

```python
graph.nodes()
```

```python
heuristic_values = {}
heuristic_values[goal] = 0
for node in graph.nodes():
    if node != goal:
        heuristic_val = int(input(f"Enter the heuristic value OF node {node}  "))
        heuristic_values[node] = heuristic_val    
```

```python
heuristic_values
```

```python
def get_heuristic(u,v):
    return heuristic_values[u]
```

```python
path = nwx.astar_path(graph, start, goal, heuristic=get_heuristic, weight='weight')
"->".join(path)
```

```python
total_cost = nwx.astar_path_length(graph, start, goal, heuristic=get_heuristic, weight='weight')
total_cost
```

```python
data = pd.DataFrame({
    "Age": np.random.randint(18, 25, 10),
    "Marks": np.random.randint(40, 100, 10),
    "Attendance": np.random.randint(60, 100, 10)
})

print("Original Dataset:\n")
print(data)

```

```python
mm = MinMaxScaler()
scaled_data = pd.DataFrame(
    mm.fit_transform(data[['Age', 'Marks', 'Attendance']]),
    columns=['Age', 'Marks', 'Attendance']
)

print("\nScaled Dataset (0 to 1 range):\n")
print(scaled_data)
```

```python
ss = StandardScaler()
scaled_data = pd.DataFrame(ss.fit_transform(data[['Age','Marks','Attendance']]))
print(scaled_data)
```
